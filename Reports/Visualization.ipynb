```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("employee_data.csv")
display(df.head())
print(df.columns)
```

1. Bar Chart

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x='City',
    y='Performance_Rating',
    hue='City',
    errorbar=None,
    legend=False
)

plt.title("Average Performance Rating by City")
plt.xlabel("City")
plt.ylabel("Average Performance Rating (Points)")

plt.tight_layout()

plt.savefig(
    "bar_chart.png",
    dpi=300
)

plt.show()
```

2. Histogram

```python
sns.histplot(
    data=df,
    x='Age',
    bins=5,
    color='#3498DB',
    edgecolor='black'
)

plt.title("Distribution of Employee Performance Ratings")
plt.xlabel("Age")
plt.ylabel("Number of Employees (Count)")

plt.tight_layout()

plt.savefig(
    "histogram.png",
    dpi=300
)

plt.show()
```

3. Scatter Plot

```python
sns.scatterplot(
    data=df,
    x='Department',
    y='Performance_Rating',
    hue='City',
    palette='Set2',
    s=100
)

plt.title("Budget vs Employee Performance Rating")
plt.xlabel("Budget (Currency Units)")
plt.ylabel("Performance Rating (Points)")
plt.legend(title="City")

plt.tight_layout()

plt.savefig(
    "scatter_plot.png",
    dpi=300
)

plt.show()
```

4. Line Graph

```python
city_performance = (
    df.groupby('City')['Performance_Rating']
    .mean()
    .sort_values()
)
plt.plot(
    city_performance.index,
    city_performance.values,
    marker='o',
    color='#E67E22',
    linewidth=2,
    markersize=8,
    label='Average Performance Rating'
)

plt.title("Average Performance Rating by City")
plt.xlabel("City")
plt.ylabel("Average Performance Rating (Points)")
plt.legend()

plt.tight_layout()

plt.savefig(
    "line_chart.png",
    dpi=300
)

plt.show()
```

5. HeatMap

```python
numeric_df = df.select_dtypes(include='number')
correlation = numeric_df.corr()

plt.figure(figsize=(9, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)
plt.title("Correlation Heatmap of Numeric Variables")
plt.xlabel("Numeric Variables")
plt.ylabel("Numeric Variables")

plt.tight_layout()

plt.savefig(
    "heatmap.png",
    dpi=300
)

plt.show()
```
